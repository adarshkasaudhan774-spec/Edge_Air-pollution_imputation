```python
# ============================================
# STEP 1.1 — Import Libraries & Define Paths
# ============================================

import pandas as pd
import numpy as np
from pathlib import Path

# Project root
PROJECT_ROOT = Path.cwd().parent

# Raw data folder
RAW_DATA = PROJECT_ROOT / "Data" / "raw data"

# Dataset paths
AIR_FILE = RAW_DATA / "air_quality_2019_2021_hourly.csv"
MET_FILE = RAW_DATA / "hourly_meteorology_16stations_2019_2021(2).csv"

print("Project Root:")
print(PROJECT_ROOT)

print("\nRaw Data Folder:")
print(RAW_DATA)

print("\nAir Quality File:")
print(AIR_FILE)

print("\nMeteorology File:")
print(MET_FILE)
```

```python
# ============================================
# STEP 1.2A — Check Files in Raw Data Folder
# ============================================

print("Files available in raw data folder:\n")

for file in RAW_DATA.iterdir():
    print(file.name)
```

```python
# ============================================
# STEP 1.2 — Load Both Datasets
# ============================================

# Correct Meteorology file path
MET_FILE = RAW_DATA / "hourly_meteorology_16stations_2019_2021.csv"

# Load Air Quality dataset
air_df = pd.read_csv(AIR_FILE)

# Load Meteorology dataset
met_df = pd.read_csv(MET_FILE)

# Display dataset shapes
print("Air Quality Dataset Shape:")
print(air_df.shape)

print("\nMeteorology Dataset Shape:")
print(met_df.shape)
```

```python
# ============================================
# STEP 1.3 — Check Columns and Sample Data
# ============================================

print("AIR QUALITY COLUMNS:")
print(air_df.columns.tolist())

print("\nAIR QUALITY SAMPLE:")
display(air_df.head())

print("\n" + "=" * 60)

print("\nMETEOROLOGY COLUMNS:")
print(met_df.columns.tolist())

print("\nMETEOROLOGY SAMPLE:")
display(met_df.head())
```

```python
# ============================================
# STEP 1.4 — Convert Timestamp & Check Time Range
# ============================================

# Convert timestamp columns to datetime
air_df["timestamp"] = pd.to_datetime(air_df["timestamp"])
met_df["timestamp"] = pd.to_datetime(met_df["timestamp"])

# Display time range
print("AIR QUALITY TIME RANGE")
print("Start:", air_df["timestamp"].min())
print("End  :", air_df["timestamp"].max())

print("\nMETEOROLOGY TIME RANGE")
print("Start:", met_df["timestamp"].min())
print("End  :", met_df["timestamp"].max())

# Number of unique timestamps
print("\nUnique timestamps:")
print("Air Quality:", air_df["timestamp"].nunique())
print("Meteorology:", met_df["timestamp"].nunique())
```

```python
# ============================================
# STEP 1.5 — Verify Stations
# ============================================

air_stations = sorted(air_df["station_id"].unique())
met_stations = sorted(met_df["station_id"].unique())

print("Air Quality Stations:")
print(air_stations)

print("\nNumber of Air Quality Stations:")
print(len(air_stations))

print("\nMeteorology Stations:")
print(met_stations)

print("\nNumber of Meteorology Stations:")
print(len(met_stations))

# Check whether both datasets have exactly the same stations
print("\nDo both datasets have the same stations?")
print(air_stations == met_stations)
```

```python
# ============================================
# STEP 1.6 — Check Duplicate Station-Timestamp
# ============================================

# Check duplicates in Air Quality dataset
air_duplicates = air_df.duplicated(
    subset=["station_id", "timestamp"]
).sum()

# Check duplicates in Meteorology dataset
met_duplicates = met_df.duplicated(
    subset=["station_id", "timestamp"]
).sum()

print("Air Quality duplicate station-timestamp records:")
print(air_duplicates)

print("\nMeteorology duplicate station-timestamp records:")
print(met_duplicates)
```

```python
# ============================================
# STEP 1.7 — Merge Air Quality + Meteorology
# ============================================

# Merge both datasets
merged_df = pd.merge(
    air_df,
    met_df,
    on=["station_id", "station_name", "timestamp"],
    how="inner",
    suffixes=("_air", "_met")
)

# Display merged dataset shape
print("Merged Dataset Shape:")
print(merged_df.shape)

# Display columns
print("\nMerged Dataset Columns:")
print(merged_df.columns.tolist())
```

```python
# ============================================
# STEP 1.8 — Verify 13 Model Features
# ============================================

# Air pollution features
pollution_features = [
    "pm25",
    "pm10",
    "no2",
    "o3",
    "so2",
    "nox",
    "co"
]

# Meteorological features
weather_features = [
    "temperature",
    "relative_humidity",
    "wind_speed",
    "wind_direction",
    "pressure",
    "rainfall"
]

# Combine all model features
feature_columns = pollution_features + weather_features

print("Air Pollution Features:")
print(pollution_features)
print("Count:", len(pollution_features))

print("\nMeteorological Features:")
print(weather_features)
print("Count:", len(weather_features))

print("\nTotal Model Features:")
print(feature_columns)
print("Total Count:", len(feature_columns))

# Verify all features exist in merged dataset
missing_features = [
    feature for feature in feature_columns
    if feature not in merged_df.columns
]

print("\nMissing Features:")
print(missing_features)

print("\n13 Features Verified:")
print(len(feature_columns) == 13 and len(missing_features) == 0)
```

# Data cleaning


```python
# ============================================
# STEP 2.1 — Check Missing Values
# ============================================

missing_counts = merged_df.isna().sum()

print("Missing Values in Each Column:")
print(missing_counts)

print("\nTotal Missing Values:")
print(missing_counts.sum())
```

```python
# ============================================
# STEP 2.2 — Calculate Missing Percentage
# ============================================

# Calculate missing percentage for each model feature
missing_percentage = (
    merged_df[feature_columns]
    .isna()
    .mean()
    * 100
)

# Create summary table
missing_summary = pd.DataFrame({
    "Missing_Count": merged_df[feature_columns].isna().sum(),
    "Missing_Percentage": missing_percentage
})

# Sort by missing count
missing_summary = missing_summary.sort_values(
    by="Missing_Count",
    ascending=False
)

print("Missing Value Summary:")
display(missing_summary)
```

```python
# ============================================
# STEP 2.3 — Station-wise Missing Values
# ============================================

# Calculate total missing values for each station
station_missing = (
    merged_df
    .groupby(["station_id", "station_name"])[feature_columns]
    .apply(lambda x: x.isna().sum().sum())
    .reset_index(name="Missing_Count")
)

# Calculate total possible feature values per station
total_values_per_station = (
    merged_df.groupby(["station_id", "station_name"]).size()
    * len(feature_columns)
)

# Add total values
station_missing["Total_Values"] = (
    station_missing.set_index(["station_id", "station_name"]).index
    .map(total_values_per_station)
)

# Calculate missing percentage
station_missing["Missing_Percentage"] = (
    station_missing["Missing_Count"]
    / station_missing["Total_Values"]
    * 100
)

# Sort by missing percentage
station_missing = station_missing.sort_values(
    "Missing_Percentage",
    ascending=False
)

display(station_missing)
```

```python
# ============================================
# STEP 2.4 — Check Completely Missing
# Feature-Station Combinations
# ============================================

# Calculate missing values for each station and feature
station_feature_missing = (
    merged_df
    .groupby(["station_id", "station_name"])[feature_columns]
    .apply(lambda x: x.isna().sum())
)

# Find station-feature combinations where ALL values are missing
completely_missing = []

for station in station_feature_missing.index:
    for feature in feature_columns:
        missing_count = station_feature_missing.loc[station, feature]
        
        if missing_count == 26304:
            completely_missing.append({
                "station_id": station[0],
                "station_name": station[1],
                "feature": feature,
                "missing_count": missing_count
            })

completely_missing_df = pd.DataFrame(completely_missing)

print("Completely Missing Station-Feature Combinations:")
display(completely_missing_df)

print("\nTotal Completely Missing Combinations:")
print(len(completely_missing_df))
```

```python
# ============================================
# STEP 2.5 — Feature-wise Missingness by Station
# ============================================

station_feature_summary = (
    merged_df
    .groupby(["station_id", "station_name"])[feature_columns]
    .apply(lambda x: x.isna().mean() * 100)
)

print("Missing Percentage (%) for Each Feature at Each Station:")

display(
    station_feature_summary.round(2)
)
```

```python
# ============================================
# STEP 2.6 — Final Data Quality Check
# ============================================

# 1. Duplicate complete rows
duplicate_rows = merged_df.duplicated().sum()

# 2. Invalid / missing timestamps
missing_timestamps = merged_df["timestamp"].isna().sum()

# 3. Infinite values in numeric feature columns
infinite_values = np.isinf(
    merged_df[feature_columns].select_dtypes(include=np.number)
).sum().sum()

print("FINAL DATA QUALITY CHECK")
print("--------------------------------")
print("Duplicate complete rows :", duplicate_rows)
print("Missing timestamps      :", missing_timestamps)
print("Infinite values         :", infinite_values)
```

# Missing-Value Masking

```python
# ============================================
# STEP 3.1 — Create Original Observation Mask
# ============================================

# 1 = value originally available
# 0 = value originally missing

original_observation_mask = (
    merged_df[feature_columns]
    .notna()
    .astype("int8")
)

print("Original Observation Mask Shape:")
print(original_observation_mask.shape)

print("\nMask Values:")
print(original_observation_mask.stack().value_counts())

print("\n1 = Originally Observed")
print("0 = Originally Missing")
```

```python
# ============================================
# STEP 3.2 — Create Artificial Missing Mask
# ============================================

np.random.seed(42)

missing_rate = 0.20

# Copy original observation mask
artificial_mask_array = original_observation_mask.to_numpy(copy=True)

# Generate random values
random_values = np.random.random(
    artificial_mask_array.shape
)

# Select only originally observed values
artificial_positions = (
    (artificial_mask_array == 1) &
    (random_values < missing_rate)
)

# Mark selected values as artificially missing
artificial_mask_array[artificial_positions] = 0

# Convert back to DataFrame
artificial_mask = pd.DataFrame(
    artificial_mask_array,
    index=merged_df.index,
    columns=feature_columns
)

print("Artificial Mask Shape:")
print(artificial_mask.shape)

print("\nArtificial Mask Values:")
print(artificial_mask.stack().value_counts())

print("\nArtificially Masked Values:")
print(artificial_positions.sum())
```

```python
# ============================================
# STEP 3.3 — Correctly Verify Artificial Mask
# ============================================

# Positions that were artificially masked
artificially_masked = (
    (original_observation_mask == 1) &
    (artificial_mask == 0)
)

# Check whether every artificially masked value
# was originally observed
invalid_artificial_masks = (
    (original_observation_mask == 0) &
    artificially_masked
)

print("Artificially Masked Values:")
print(artificially_masked.to_numpy().sum())

print("\nInvalid Artificial Masks:")
print(invalid_artificial_masks.to_numpy().sum())

print("\nMask Verification:")

if invalid_artificial_masks.to_numpy().sum() == 0:
    print("PASS")
else:
    print("FAIL")
```

```python
# ============================================
# STEP 3.4 — Save Fixed Artificial Mask
# ============================================

PROCESSED_DATA = PROJECT_ROOT / "Data" / "processed"

# Create processed folder if it does not exist
PROCESSED_DATA.mkdir(parents=True, exist_ok=True)

# Save the fixed artificial mask
MASK_FILE = PROCESSED_DATA / "fixed_artificial_mask_new.csv"

artificial_mask.to_csv(
    MASK_FILE,
    index=False
)

print("Fixed Artificial Mask Saved Successfully!")
print("\nFile:")
print(MASK_FILE)

print("\nShape:")
print(artificial_mask.shape)
```

# Sliding-Window Dataset Creation

```python
# ============================================
# STEP 4.1 — Create Data Tensor
# ============================================

# Sort data properly
merged_df = merged_df.sort_values(
    ["timestamp", "station_id"]
).reset_index(drop=True)

# Get ordered timestamps and stations
timestamps = sorted(merged_df["timestamp"].unique())
stations = sorted(merged_df["station_id"].unique())

# Create 3D data tensor
data_tensor = np.full(
    (len(timestamps), len(stations), len(feature_columns)),
    np.nan,
    dtype=np.float32
)

# Fill tensor
for t_idx, timestamp in enumerate(timestamps):
    timestamp_data = merged_df[
        merged_df["timestamp"] == timestamp
    ].sort_values("station_id")

    data_tensor[t_idx, :, :] = (
        timestamp_data[feature_columns]
        .to_numpy(dtype=np.float32)
    )

print("Data Tensor Shape:")
print(data_tensor.shape)

print("\nExpected Shape:")
print("(26304, 16, 13)")

print("\nShape Verification:")
print(data_tensor.shape == (26304, 16, 13))
```

```python
# ============================================
# STEP 4.2 — Create Mask Tensor
# ============================================

# Convert fixed artificial mask into numpy array
mask_array = artificial_mask.to_numpy(
    dtype=np.int8
)

# Create mask tensor
mask_tensor = np.full(
    (len(timestamps), len(stations), len(feature_columns)),
    0,
    dtype=np.int8
)

# Fill mask tensor
for t_idx, timestamp in enumerate(timestamps):
    timestamp_mask = merged_df[
        merged_df["timestamp"] == timestamp
    ].sort_values("station_id")

    mask_tensor[t_idx, :, :] = (
        artificial_mask.loc[
            timestamp_mask.index,
            feature_columns
        ]
        .to_numpy(dtype=np.int8)
    )

print("Mask Tensor Shape:")
print(mask_tensor.shape)

print("\nExpected Shape:")
print("(26304, 16, 13)")

print("\nShape Verification:")
print(mask_tensor.shape == (26304, 16, 13))

print("\nMask Values:")
print(
    np.unique(
        mask_tensor,
        return_counts=True
    )
)
```

```python
# ============================================
# STEP 4.3 — Create 24-Hour Sliding Windows
# ============================================

WINDOW_SIZE = 24

# Create data sliding windows
data_windows = np.array([
    data_tensor[i:i + WINDOW_SIZE]
    for i in range(
        len(data_tensor) - WINDOW_SIZE + 1
    )
])

# Create corresponding mask sliding windows
mask_windows = np.array([
    mask_tensor[i:i + WINDOW_SIZE]
    for i in range(
        len(mask_tensor) - WINDOW_SIZE + 1
    )
])

print("Data Windows Shape:")
print(data_windows.shape)

print("\nMask Windows Shape:")
print(mask_windows.shape)

print("\nExpected Shape:")
print("(26281, 24, 16, 13)")

print("\nData Shape Verification:")
print(data_windows.shape == (26281, 24, 16, 13))

print("\nMask Shape Verification:")
print(mask_windows.shape == (26281, 24, 16, 13))
```

```python
# ============================================
# STEP 4.4 — Final Sliding-Window Verification
# ============================================

print("DATA WINDOWS")
print("Shape:", data_windows.shape)
print("NaN Count:", np.isnan(data_windows).sum())

print("\nMASK WINDOWS")
print("Shape:", mask_windows.shape)
print("Unique Values:", np.unique(mask_windows))

print("\nALIGNMENT CHECK")
print(
    "Same Number of Windows:",
    data_windows.shape[0] == mask_windows.shape[0]
)

print(
    "Same Time Window:",
    data_windows.shape[1] == mask_windows.shape[1]
)

print(
    "Same Stations:",
    data_windows.shape[2] == mask_windows.shape[2]
)

print(
    "Same Features:",
    data_windows.shape[3] == mask_windows.shape[3]
)
```
