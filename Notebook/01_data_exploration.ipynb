# Missing-Value Masking

```python
import pandas as pd

file_path = "../Data/raw data/air_weather_hourly_2019_2021 (2).csv"

df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
# STEP 3.1 — Check original missing values

station_columns = [
    "SHATIN", "TSUEN WAN", "CENTRAL", "EASTERN",
    "KWUN TONG", "TUEN MUN", "TUNG CHUNG", "SHAM SHUI PO",
    "SOUTHERN", "YUEN LONG", "CENTRAL/WESTERN", "KWAI CHUNG",
    "TSEUNG KWAN O", "TAI PO", "MONG KOK", "CAUSEWAY BAY"
]

original_missing = df[station_columns].isna().sum()

print("Missing values in each station:")
print(original_missing)

print("\nTotal original missing values:", original_missing.sum())

print("\nTotal station values:", df[station_columns].size)

print(
    "\nOriginal missing percentage:",
    round((original_missing.sum() / df[station_columns].size) * 100, 2),
    "%"
)
```

```python
# STEP 3.2 — Create Observed-Value Mask

# 1 = value available (observed)
# 0 = value missing

observed_mask = df[station_columns].notna().astype("int8")

print("Observed-value mask created successfully!")

print("\nMask shape:", observed_mask.shape)

print("\nFirst 5 rows of mask:")
print(observed_mask.head())

print("\nMask values:")
print(observed_mask.stack().value_counts())
```

```python
# STEP 3.3 — Verify Observed Mask

# Check:
# Mask = 0 should correspond to NaN
# Mask = 1 should correspond to available value

mask_zero = (observed_mask == 0)
mask_one = (observed_mask == 1)

# Values that are missing but mask says 1
error_missing = df[station_columns].isna() & mask_one

# Values that are available but mask says 0
error_observed = df[station_columns].notna() & mask_zero

print("Missing values incorrectly marked as 1:", error_missing.sum().sum())
print("Observed values incorrectly marked as 0:", error_observed.sum().sum())

if error_missing.sum().sum() == 0 and error_observed.sum().sum() == 0:
    print("\nMask verification successful! ✅")
else:
    print("\nMask verification failed! ❌")
```

```python
# STEP 3.4 — Create Artificial Missing Mask

import numpy as np
import pandas as pd

# Reproducible result
np.random.seed(42)

# Artificial missingness rate
missing_rate = 0.20

# Mask ko writable NumPy array mein convert karo
mask_array = observed_mask.to_numpy(copy=True)

# Random numbers generate karo
random_values = np.random.random(mask_array.shape)

# Sirf originally observed values mein se 20% select karo
artificial_positions = (
    (mask_array == 1) &
    (random_values < missing_rate)
)

# Selected values ko artificially missing mark karo
mask_array[artificial_positions] = 0

# NumPy array ko wapas DataFrame mein convert karo
artificial_mask = pd.DataFrame(
    mask_array,
    index=observed_mask.index,
    columns=observed_mask.columns
)

# Results
original_observed = (observed_mask.to_numpy() == 1).sum()
artificially_masked = artificial_positions.sum()

print("Artificial masking completed successfully! ✅")

print("\nOriginal observed values:", original_observed)
print("Artificially masked values:", artificially_masked)

print(
    "Actual artificial missing rate:",
    round((artificially_masked / original_observed) * 100, 2),
    "%"
)

print("\nFinal mask counts:")
print(artificial_mask.stack().value_counts())
```

 # Sliding-Window Dataset

```python
# STEP 4.0 — Restore Dataset and Mask Variables

import pandas as pd
import numpy as np

# Dataset load
file_path = "../Data/raw data/air_weather_hourly_2019_2021 (2).csv"
df = pd.read_csv(file_path)

# 16 station columns
station_columns = [
    "SHATIN", "TSUEN WAN", "CENTRAL", "EASTERN",
    "KWUN TONG", "TUEN MUN", "TUNG CHUNG", "SHAM SHUI PO",
    "SOUTHERN", "YUEN LONG", "CENTRAL/WESTERN", "KWAI CHUNG",
    "TSEUNG KWAN O", "TAI PO", "MONG KOK", "CAUSEWAY BAY"
]

# Original observed mask
observed_mask = df[station_columns].notna().astype("int8")

# Recreate the same 20% artificial mask
np.random.seed(42)

missing_rate = 0.20

mask_array = observed_mask.to_numpy(copy=True)

random_values = np.random.random(mask_array.shape)

artificial_positions = (
    (mask_array == 1) &
    (random_values < missing_rate)
)

mask_array[artificial_positions] = 0

artificial_mask = pd.DataFrame(
    mask_array,
    index=observed_mask.index,
    columns=observed_mask.columns
)

print("Dataset and mask variables restored successfully! ✅")

print("\nDataset shape:", df.shape)
print("Observed mask shape:", observed_mask.shape)
print("Artificial mask shape:", artificial_mask.shape)

print("\nArtificially masked values:",
      artificial_positions.sum())
```

```python
# STEP 4.1 — Check Time and Pollutant Structure

print("Total rows:", len(df))

print("\nUnique timestamps:", df["TIMESTAMP"].nunique())

print("\nUnique pollutants:")
print(df["POLLUTANT"].unique())

print("\nNumber of pollutants:", df["POLLUTANT"].nunique())

print("\nFirst 10 timestamps:")
print(df["TIMESTAMP"].head(10))

print("\nRows per timestamp:")
print(df.groupby("TIMESTAMP").size().value_counts().sort_index())
```

```python
# STEP 4.2 — Create Time × Station × Feature Tensor

import numpy as np
import pandas as pd

# --------------------------------------------------
# 1. Pollution features
# --------------------------------------------------

pollutant_order = [
    "Sulphur Dioxide",
    "Respirable Suspended Particulates",
    "Fine Suspended Particulates ",
    "Nitrogen Oxides",
    "Nitrogen Dioxide",
    "Carbon Monoxide",
    "Ozone"
]

# --------------------------------------------------
# 2. Weather features
# --------------------------------------------------

weather_columns = [
    "Temperature_mean",
    "Humidity_mean",
    "Sea_Level_Pressure_mean",
    "Wind_Speed_mean",
    "Available_Stations",
    "Wind_Direction_mean"
]

# --------------------------------------------------
# 3. Sort data by timestamp
# --------------------------------------------------

df["TIMESTAMP"] = pd.to_datetime(df["TIMESTAMP"])

df = df.sort_values("TIMESTAMP").reset_index(drop=True)

# Unique timestamps
timestamps = df["TIMESTAMP"].drop_duplicates().sort_values().reset_index(drop=True)

# --------------------------------------------------
# 4. Create pollution tensor
# Shape = (time, station, pollution_features)
# --------------------------------------------------

pollution_data = np.zeros(
    (len(timestamps), len(station_columns), len(pollutant_order)),
    dtype=np.float32
)

for feature_index, pollutant in enumerate(pollutant_order):

    pollutant_df = (
        df[df["POLLUTANT"] == pollutant]
        .sort_values("TIMESTAMP")
    )

    pollution_data[:, :, feature_index] = (
        pollutant_df[station_columns].to_numpy(dtype=np.float32)
    )

# --------------------------------------------------
# 5. Create weather data
# One row per timestamp
# --------------------------------------------------

weather_df = (
    df[["TIMESTAMP"] + weather_columns]
    .drop_duplicates("TIMESTAMP")
    .sort_values("TIMESTAMP")
)

weather_data = weather_df[weather_columns].to_numpy(
    dtype=np.float32
)

# --------------------------------------------------
# 6. Repeat weather features for all 16 stations
# --------------------------------------------------

weather_data = np.repeat(
    weather_data[:, np.newaxis, :],
    len(station_columns),
    axis=1
)

# --------------------------------------------------
# 7. Combine pollution + weather
# --------------------------------------------------

data_tensor = np.concatenate(
    [pollution_data, weather_data],
    axis=2
)

# --------------------------------------------------
# 8. Display final shape
# --------------------------------------------------

print("Tensor created successfully! ✅")

print("\nTensor shape:", data_tensor.shape)

print("\nExpected shape:")
print("(26304, 16, 13)")

print("\nNumber of timestamps:", data_tensor.shape[0])
print("Number of stations:", data_tensor.shape[1])
print("Number of features:", data_tensor.shape[2])
```

```python
# STEP 4.3 — Create 24-Hour Sliding Windows

window_size = 24

num_windows = data_tensor.shape[0] - window_size + 1

sliding_windows = np.empty(
    (num_windows, window_size, len(station_columns), 13),
    dtype=np.float32
)

for i in range(num_windows):
    sliding_windows[i] = data_tensor[
        i:i + window_size
    ]

print("Sliding-window dataset created successfully! ✅")

print("\nWindow size:", window_size, "hours")

print("Number of windows:", sliding_windows.shape[0])

print("Each window shape:", sliding_windows.shape[1:])

print("Final sliding-window shape:", sliding_windows.shape)

print(
    "\nExpected shape:",
    (26304 - 24 + 1, 24, 16, 13)
)
```

```python
# STEP 4.4 — Create 13-Feature Mask Tensor

# Pollution mask:
# 7 pollutant features ke liye artificial_mask use hoga

pollution_mask = np.zeros(
    (len(timestamps), len(station_columns), len(pollutant_order)),
    dtype=np.int8
)

for feature_index, pollutant in enumerate(pollutant_order):

    pollutant_df = (
        df[df["POLLUTANT"] == pollutant]
        .sort_values("TIMESTAMP")
    )

    # Original missing + artificial missing
    pollution_mask[:, :, feature_index] = (
        artificial_mask.loc[
            pollutant_df.index, station_columns
        ].to_numpy(dtype=np.int8)
    )


# Weather mask:
# Weather features mein missing values nahi hain,
# isliye sabko 1 mark karenge.

weather_mask = np.ones(
    (len(timestamps), len(station_columns), len(weather_columns)),
    dtype=np.int8
)


# Combine pollution + weather masks

mask_tensor = np.concatenate(
    [pollution_mask, weather_mask],
    axis=2
)


print("13-feature mask tensor created successfully! ✅")

print("\nMask tensor shape:", mask_tensor.shape)

print("\nExpected shape:")
print("(26304, 16, 13)")

print("\nMask values:")
print(
    pd.Series(mask_tensor.flatten()).value_counts().sort_index()
)
```

```python
# STEP 4.5 — Create Sliding-Window Mask

mask_sliding_windows = np.empty(
    (
        num_windows,
        window_size,
        len(station_columns),
        13
    ),
    dtype=np.int8
)

for i in range(num_windows):
    mask_sliding_windows[i] = mask_tensor[
        i:i + window_size
    ]

print("Sliding-window mask created successfully! ✅")

print("\nMask window size:", window_size, "hours")

print("Number of mask windows:", mask_sliding_windows.shape[0])

print("Each mask window shape:", mask_sliding_windows.shape[1:])

print(
    "Final mask sliding-window shape:",
    mask_sliding_windows.shape
)

print(
    "\nData windows shape:",
    sliding_windows.shape
)

print(
    "Data and mask shapes match:",
    sliding_windows.shape == mask_sliding_windows.shape
)
```
